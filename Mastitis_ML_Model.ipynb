```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Load clinical dataset
print("Loading bovine mastitis dataset...")
df = pd.read_csv('cow_milk_mastitis_dataset.csv')

# 2. Define Features (X) and Target (y)
# class1: 0 = Healthy, 1 = Mastitis Risk
X = df[['Milk_Conductivity', 'Milk_Temperature']]
y = df['class1']

# 3. Train Logistic Regression Model
print("Training TinyML Logistic Regression Engine...")
model = LogisticRegression()
model.fit(X, y)

# 4. Evaluation
predictions = model.predict(X)
acc = accuracy_score(y, predictions)
print(f"Model Training Accuracy: {acc * 100:.2f}%\n")

# 5. Output C++ Weights for Arduino config.h
print("==================================================")
print("  PASTE THESE EXACT WEIGHTS INTO ARDUINO config.h ")
print("==================================================")
print(f"const float W_BIAS = {model.intercept_[0]:.2f};")
print(f"const float W_EC   = {model.coef_[0][0]:.2f};")
print(f"const float W_TEMP = {model.coef_[0][1]:.2f};")
print("==================================================")
```
